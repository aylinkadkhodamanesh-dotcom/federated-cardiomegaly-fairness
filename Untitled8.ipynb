```python
!pip install -q medmnist
import numpy as np, pandas as pd, matplotlib.pyplot as plt
import torch, torch.nn as nn, copy, time
from sklearn.metrics import roc_auc_score
from medmnist import ChestMNIST, INFO

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Running on:", device)
print("Label 1 =", INFO['chestmnist']['label']['1'])
```

```python
train = ChestMNIST(split='train', download=True, size=28)
test  = ChestMNIST(split='test',  download=True, size=28)

for name, ds in [('train', train), ('test', test)]:
    y = ds.labels[:, 1]
    print(f"{name}: {len(ds)} images, cardiomegaly {y.sum()} ({y.mean()*100:.1f}%)")
```

```python
rng = np.random.default_rng(42)
X_tr = train.imgs.astype(np.float32) / 255.0
y_tr = train.labels[:, 1].astype(np.float32)
X_te = test.imgs.astype(np.float32) / 255.0
y_te = test.labels[:, 1].astype(np.float32)

# (number of training images, cardiomegaly rate)
plan = {
    "H1_large":  (30000, 0.040),
    "H2_medium": (20000, 0.020),
    "H3_small":  (8000,  0.030),
    "H4_tiny":   (3000,  0.015),
}

def scanner(x, h, seed=0):
    if h == "H1_large":
        return x
    if h == "H2_medium":
        return x ** 0.7                                   # brighter
    if h == "H3_small":
        return np.clip(0.5 + 0.6 * (x - 0.5), 0, 1)       # lower contrast
    if h == "H4_tiny":
        r = np.random.default_rng(seed)
        return np.clip(x ** 1.4 + r.normal(0, 0.05, x.shape), 0, 1)  # darker and noisy

pos = rng.permutation(np.where(y_tr == 1)[0])
neg = rng.permutation(np.where(y_tr == 0)[0])
test_chunks = np.array_split(rng.permutation(len(y_te)), 4)

hospitals = {}
p_i = n_i = 0
for k, (h, (n, prev)) in enumerate(plan.items()):
    n_pos = int(n * prev)
    idx = np.concatenate([pos[p_i:p_i + n_pos], neg[n_i:n_i + n - n_pos]])
    p_i += n_pos; n_i += n - n_pos
    t = test_chunks[k]
    hospitals[h] = dict(
        X=scanner(X_tr[idx], h, seed=k).astype(np.float32), y=y_tr[idx],
        Xt=scanner(X_te[t], h, seed=100 + k).astype(np.float32), yt=y_te[t])
names = list(hospitals)

for h, d in hospitals.items():
    print(f"{h:10s} train {len(d['y']):6d} | cardiomegaly {int(d['y'].sum()):5d} "
          f"({d['y'].mean()*100:.1f}%) | test {len(d['yt'])}, positives {int(d['yt'].sum())}")

fig, ax = plt.subplots(1, 4, figsize=(12, 3))
for a, (h, d) in zip(ax, hospitals.items()):
    a.imshow(d['X'][np.where(d['y'] == 1)[0][0]], cmap='gray', vmin=0, vmax=1)
    a.set_title(h); a.axis('off')
plt.suptitle("One cardiomegaly case per hospital")
plt.tight_layout(); plt.savefig("hospitals.png", dpi=150); plt.show()
```

```python
torch.manual_seed(42)

def loader(X, y, shuffle=True, bs=256):
    ds = torch.utils.data.TensorDataset(torch.from_numpy(X).unsqueeze(1), torch.from_numpy(y))
    return torch.utils.data.DataLoader(ds, batch_size=bs, shuffle=shuffle)

def block(cin, cout):
    return nn.Sequential(
        nn.Conv2d(cin, cout, 3, padding=1), nn.GroupNorm(8, cout), nn.ReLU(),
        nn.Conv2d(cout, cout, 3, padding=1), nn.GroupNorm(8, cout), nn.ReLU(),
        nn.MaxPool2d(2))

class SmallCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            block(1, 16), block(16, 32), block(32, 64),
            nn.AdaptiveAvgPool2d(1), nn.Flatten(), nn.Dropout(0.3), nn.Linear(64, 1))
    def forward(self, x):
        return self.net(x).squeeze(1)

def train_local(model, X, y, epochs=1, lr=1e-3):
    model.train()
    pw = torch.tensor((1 - y.mean()) / y.mean(), device=device)   # weight the rare positives
    lossf = nn.BCEWithLogitsLoss(pos_weight=pw)
    opt = torch.optim.Adam(model.parameters(), lr=lr)
    for _ in range(epochs):
        for xb, yb in loader(X, y):
            xb, yb = xb.to(device), yb.to(device)
            opt.zero_grad(); lossf(model(xb), yb).backward(); opt.step()
    return model

@torch.no_grad()
def predict(model, X):
    model.eval()
    dummy = np.zeros(len(X), np.float32)
    return torch.cat([torch.sigmoid(model(xb.to(device))).cpu()
                      for xb, _ in loader(X, dummy, shuffle=False, bs=1024)]).numpy()

def auc(model, X, y):
    return roc_auc_score(y, predict(model, X))

def eval_all(model):
    return {h: auc(model, d['Xt'], d['yt']) for h, d in hospitals.items()}

def fedavg(states, weights):
    total = sum(weights)
    avg = copy.deepcopy(states[0])
    for k in avg:
        avg[k] = sum(st[k] * (w / total) for st, w in zip(states, weights))
    return avg

EPOCHS = 8
results = {}   # results[method][hospital] = test AUC
curves = {}    # mean AUC across hospitals after each epoch or round
print("Ready")
```

```python
t0 = time.time()
results['Local only'] = {}
for h in names:
    torch.manual_seed(42)
    m = SmallCNN().to(device)
    train_local(m, hospitals[h]['X'], hospitals[h]['y'], epochs=EPOCHS)
    results['Local only'][h] = auc(m, hospitals[h]['Xt'], hospitals[h]['yt'])
    print(f"Local only | {h}: AUC {results['Local only'][h]:.3f}  ({time.time()-t0:.0f}s)")
```

```python
torch.manual_seed(42)
X_all = np.concatenate([d['X'] for d in hospitals.values()])
y_all = np.concatenate([d['y'] for d in hospitals.values()])

central = SmallCNN().to(device)
curves['Centralized'] = []
t0 = time.time()
for ep in range(EPOCHS):
    train_local(central, X_all, y_all, epochs=1)
    s = eval_all(central)
    curves['Centralized'].append(np.mean(list(s.values())))
    print(f"Centralized | epoch {ep+1}: mean AUC {curves['Centralized'][-1]:.3f}  ({time.time()-t0:.0f}s)")
results['Centralized'] = s
```

```python
torch.manual_seed(42)
global_model = SmallCNN().to(device)
curves['FedAvg'] = []
t0 = time.time()
for r in range(EPOCHS):
    states, sizes = [], []
    for h in names:                                   # each hospital trains locally
        local = copy.deepcopy(global_model)
        train_local(local, hospitals[h]['X'], hospitals[h]['y'], epochs=1)
        states.append(local.state_dict()); sizes.append(len(hospitals[h]['y']))
    global_model.load_state_dict(fedavg(states, sizes))  # server averages weights only
    s = eval_all(global_model)
    curves['FedAvg'].append(np.mean(list(s.values())))
    print(f"FedAvg | round {r+1}: mean AUC {curves['FedAvg'][-1]:.3f}, "
          f"worst site {min(s.values()):.3f}  ({time.time()-t0:.0f}s)")
results['FedAvg'] = s
```

```python
torch.manual_seed(42)
fair_model = SmallCNN().to(device)
curves['FedAvg equal weights'] = []
t0 = time.time()
for r in range(EPOCHS):
    states = []
    for h in names:
        local = copy.deepcopy(fair_model)
        train_local(local, hospitals[h]['X'], hospitals[h]['y'], epochs=1)
        states.append(local.state_dict())
    fair_model.load_state_dict(fedavg(states, [1] * len(names)))   # every hospital counts equally
    s = eval_all(fair_model)
    curves['FedAvg equal weights'].append(np.mean(list(s.values())))
    print(f"FedAvg equal weights | round {r+1}: mean AUC {curves['FedAvg equal weights'][-1]:.3f}, "
          f"worst site {min(s.values()):.3f}  ({time.time()-t0:.0f}s)")
results['FedAvg equal weights'] = s
```

```python
df = pd.DataFrame(results).T[names]
df['Mean'] = df[names].mean(axis=1)
df['Worst site'] = df[names].min(axis=1)
df['Gap (best - worst)'] = df[names].max(axis=1) - df[names].min(axis=1)
print(df.round(3).to_string())
df.round(3).to_csv("results.csv")

fig, ax = plt.subplots(figsize=(9, 4))
w = 0.8 / len(df)
x = np.arange(len(names))
for i, (method, row) in enumerate(df.iterrows()):
    ax.bar(x + i * w, row[names], w, label=method)
ax.set_xticks(x + w * (len(df) - 1) / 2); ax.set_xticklabels(names)
ax.set_ylim(0.5, 1); ax.set_ylabel("Test AUC")
ax.set_title("Cardiomegaly detection: AUC per hospital"); ax.legend(fontsize=8)
plt.tight_layout(); plt.savefig("auc_per_hospital.png", dpi=150); plt.show()

fig, ax = plt.subplots(figsize=(6, 4))
for k, v in curves.items():
    ax.plot(range(1, len(v) + 1), v, marker='o', label=k)
ax.axhline(df.loc['Local only', 'Mean'], ls='--', color='gray', label='Local only (mean)')
ax.set_xlabel("Epoch / round"); ax.set_ylabel("Mean AUC across hospitals")
ax.set_title("Federated vs centralized training"); ax.legend(fontsize=8)
plt.tight_layout(); plt.savefig("learning_curves.png", dpi=150); plt.show()
```
